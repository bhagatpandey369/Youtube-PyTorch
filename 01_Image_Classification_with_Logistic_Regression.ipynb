```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
```

```python
train_dataset = datasets.FashionMNIST(root='/dataset', train=True, transform=transforms.ToTensor(), download=True)
```

```python
train_dataset
```

```python
test_dataset = datasets.FashionMNIST(root='/dataset', train=False, transform=transforms.ToTensor())
```

```python
test_dataset
```

```python
len(train_dataset), len(test_dataset)
```

```python
batch_size = 64
```

```python
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python
len(train_loader)
```

```python
len(test_loader)
```

```python
class LogisticRegression(nn.Module):
  def __init__(self, in_dim, n_class):
    super().__init__()
    self.linear = nn.Linear(in_dim, n_class)

  def forward(self, x):
    x = self.linear(x)
    return x
```

```python
model = LogisticRegression(28*28, 10)
```

```python
model
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
device
```

```python
model = model.to(device)
```

```python
learning_rate = 0.001
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
```

```python
num_epoch = 10
```

```python
for epoch in range(num_epoch):
  print(epoch+1,'*'*20)
  running_loss = 0.
  running_acc = 0.
  model.train()
  for i, data in enumerate(train_loader, 1):
    break


```
